## ¿Por qué NumPy?

Python "puro" puede trabajar con listas de números, pero cuando tenemos **muchos** datos numéricos
se vuelve lento y engorroso. **NumPy** introduce el `array` (arreglo): una estructura pensada para
guardar y operar sobre grandes cantidades de números **de forma muy rápida**.

Casi todas las librerías de datos de Python (incluida Pandas) usan NumPy por debajo.

### 1️⃣ Importar NumPy

Por convención se importa con el alias `np`.

````python
import numpy as np

print("NumPy versión:", np.__version__)
````

### 2️⃣ Crear un arreglo

Un arreglo se crea con `np.array()` a partir de una lista de Python.
Imaginemos que tenemos los **valores de vivienda** (en dólares) de 5 distritos:

````python
valores = np.array([452600, 358500, 352100, 341300, 342200])
print(valores)
print(type(valores))
````

### 3️⃣ Consultar sus propiedades

Todo arreglo tiene atributos que describen su **estructura**:

- `shape` → la forma (cuántas filas y columnas).
- `size`  → el número total de elementos.
- `ndim`  → el número de dimensiones (1 = vector, 2 = matriz, ...).

````python
print("shape:", valores.shape)   # (5,) -> 5 elementos en una dimensión
print("size :", valores.size)    # 5
print("ndim :", valores.ndim)    # 1
````

### 4️⃣ Operaciones básicas

Aquí está la magia de NumPy: podemos operar sobre **todo el arreglo a la vez**,
sin escribir bucles. Esto se llama **vectorización**.

````python
# Operaciones aritméticas elemento a elemento
print("Valores + 10000:", valores + 10000)   # le suma 10000 a CADA elemento
print("Valores / 1000  :", valores / 1000)    # los pasa a miles de dólares
````

````python
# Operaciones estadísticas sobre todo el arreglo
print("Suma    :", valores.sum())
print("Promedio:", valores.mean())
print("Máximo  :", valores.max())
print("Mínimo  :", valores.min())
````

### 5️⃣ Reflexión: ¿por qué no una lista normal?

Con una lista de Python, sumarle 10000 a cada elemento requeriría un bucle:

```python
# Con lista normal (más largo y más lento con muchos datos)
valores_lista = [452600, 358500, 352100, 341300, 342200]
resultado = [v + 10000 for v in valores_lista]
```

Con NumPy es simplemente `valores + 10000`. Más corto, más legible y **mucho más rápido**
cuando hay millones de datos.

> 💡 **Conexión con el taller:** más adelante, cuando carguemos el dataset con Pandas,
> cada columna numérica es en realidad un arreglo de NumPy por debajo. Por eso podremos
> calcular promedios, máximos y mínimos con la misma facilidad.
