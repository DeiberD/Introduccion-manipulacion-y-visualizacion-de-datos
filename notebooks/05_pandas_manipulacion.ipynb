## Ahora sí: hacerle preguntas a los datos

Ya cargamos y exploramos el dataset. En esta actividad aprenderemos a **manipularlo** para
obtener exactamente la información que necesitamos. Volvemos a cargarlo para trabajar:

```python
import pandas as pd

df = pd.read_csv("../data/housing.csv")
print("Dataset cargado:", df.shape)
```

### 1. Seleccionar columnas

Para quedarnos solo con las columnas que nos interesan, usamos una **lista de nombres**
entre corchetes dobles `[[ ]]`.

```python
# Una sola columna (devuelve una Series)
df["median_house_value"].head()
```

```python
# Varias columnas (devuelve un DataFrame) -> ¡ojo con los corchetes dobles!
df[["median_income", "median_house_value", "ocean_proximity"]].head()
```

### 2. Filtrar registros con una condición

Podemos quedarnos solo con las filas que cumplen una condición. Por ejemplo, los distritos
con viviendas **muy caras** (valor mayor a 450.000 USD):

```python
caras = df[df["median_house_value"] > 450000]
print("Distritos con viviendas > 450.000 USD:", caras.shape[0])
caras.head()
```

La condición `df["median_house_value"] > 450000` produce una Series de `True`/`False`,
y Pandas nos devuelve solo las filas donde el resultado es `True`.

### 3. Ordenar los datos con `sort_values()`

Ordenemos el dataset por valor de vivienda, de mayor a menor (`ascending=False`):

```python
df.sort_values("median_house_value", ascending=False).head()
```

### 4. Detectar datos nulos

¿Recuerdas que `total_bedrooms` tenía datos faltantes? Podemos contarlos con
`isnull()` (marca cada celda vacía como `True`) combinado con `sum()`:

```python
# Nulos por columna
df.isnull().sum()
```

Vemos **207 valores nulos** en `total_bedrooms`. En un análisis real habría que decidir
qué hacer con ellos (eliminarlos, rellenarlos con un promedio, etc.). Por hoy solo nos interesa
**saber detectarlos**.

### 5. Agrupar con `groupby()`

`groupby()` es una de las herramientas más poderosas de Pandas. Permite **partir los datos en grupos**
y calcular una medida para cada grupo. Con esto respondemos nuestra **pregunta 2**:

> *¿La cercanía al océano influye en el valor de las viviendas?*

Agrupamos por `ocean_proximity` y calculamos el valor **promedio** de vivienda de cada grupo:

```python
valor_por_zona = df.groupby("ocean_proximity")["median_house_value"].mean().round(0)
valor_por_zona.sort_values(ascending=False)
```

Un promedio, por sí solo, no dice cuántos datos lo respaldan. Con `agg()` podemos pedir **varias
medidas a la vez**: el promedio (`mean`) y la cantidad de distritos (`count`) de cada grupo.

```python
df.groupby("ocean_proximity")["median_house_value"].agg(["mean", "count"]).round(0)
```

### Interpretación

¡Los datos ya nos están hablando! Las viviendas cerca de la bahía (`NEAR BAY`) u océano
(`NEAR OCEAN`, `<1H OCEAN`) valen bastante más que las de tierra adentro (`INLAND`).

¿Y `ISLAND`, que aparece como la más cara? Recuerda el **momento detective** de la actividad
anterior: ese promedio sale de **solo 5 distritos**, mientras que los demás grupos tienen miles.
Es un dato curioso, pero no una conclusión confiable. **Un promedio sin contexto puede engañar.**

Aun así, la evidencia de los grupos grandes sugiere que **sí**, la cercanía al océano parece influir
en el valor. Lo confirmaremos con una gráfica en la actividad de Seaborn.
