## Respondiendo la pregunta guía

> **"¿De qué depende el valor de una vivienda en California?"**

Encontramos (al menos) dos factores importantes:

1. **El ingreso de la zona** → a mayor `median_income`, mayor valor de vivienda
   (lo vimos en el scatter, con tendencia ascendente).
2. **La cercanía al océano** → las zonas costeras valen más que las de tierra adentro
   (lo vimos en el `groupby` y en el `boxplot`).

**¿Y la apuesta del inicio?** Si elegiste **A (ingreso)** o **B (océano)**, acertaste: son los dos
factores con más peso. La antigüedad (C) y el número de habitaciones (D) casi no se relacionan con
el valor. A veces la intuición acierta... y a veces no; por eso miramos los datos.

**Importante:** "relación" no es lo mismo que "causa". Los datos muestran una **asociación**,
pero para afirmar causalidad haría falta un análisis más profundo. ¡Buen tema para seguir explorando!

Pistas:

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/housing.csv")

# Tu código aquí. Pista: un scatterplot de 'housing_median_age' vs 'median_house_value'.
````

<details>
<summary><b>Ver una posible solución</b> (inténtalo primero)</summary>

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="housing_median_age", y="median_house_value", alpha=0.2, s=10)
plt.title("Antigüedad vs. valor de la vivienda")
plt.xlabel("Antigüedad media (años)")
plt.ylabel("Valor medio de la vivienda (USD)")
plt.show()
```

Los puntos se reparten casi por igual a lo largo de todas las edades: **no hay una relación clara**
entre la antigüedad y el valor (la correlación es apenas 0,11). Un buen recordatorio de que no
todas las variables importan, y de que también es un hallazgo descubrir que algo **no** influye.
