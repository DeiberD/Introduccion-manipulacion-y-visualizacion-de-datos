## De un archivo a un DataFrame

Los datos casi nunca los escribimos a mano: vienen en archivos. El formato más común es el
**`.csv`** (valores separados por comas). Pandas lo lee con una sola función: `read_csv()`.

```python
import pandas as pd

# Cargamos el dataset del taller.
# La ruta '../data/housing.csv' significa: "sube una carpeta y entra a 'data'".
df = pd.read_csv("../data/housing.csv")

print("¡Dataset cargado!")
print("Forma:", df.shape, "-> (filas, columnas)")
```

### 1. `head()` — los primeros registros

`head()` nos muestra las primeras filas (5 por defecto). Sirve para echarle un primer vistazo
a los datos y ver qué "pinta" tienen.

```python
df.head()
```

### 2. `info()` — la información general

`info()` nos dice, para cada columna: su **nombre**, cuántos valores **no nulos** tiene y su
**tipo de dato** (`float64` = número decimal, `object`/`str` = texto).

**Fíjate** especialmente en la columna `total_bedrooms`: tiene **menos** valores no nulos que las
demás. ¡Eso significa que le faltan datos! Lo veremos en la siguiente actividad.

```python
df.info()
```

### 3. `describe()` — estadísticas descriptivas

`describe()` calcula automáticamente estadísticas de las columnas **numéricas**: cuántos datos hay,
el promedio (`mean`), la desviación estándar (`std`), el mínimo, el máximo y los percentiles.

```python
df.describe()
```

### 4. `value_counts()` — ¿cuántos hay de cada categoría?

`describe()` solo resume las columnas **numéricas**. Para una columna de texto (categórica) como
`ocean_proximity`, usamos `value_counts()`, que cuenta cuántas veces aparece cada valor:

```python
df["ocean_proximity"].value_counts()
```

**Momento detective:** la categoría `ISLAND` tiene apenas **5 distritos**, frente a miles en las
demás. Guarda este dato en la memoria: más adelante nos va a evitar sacar una conclusión equivocada.

### 5. ¿Qué significan las columnas?

| Columna | Significado |
|---------|-------------|
| `longitude`, `latitude` | Ubicación geográfica del distrito |
| `housing_median_age` | Antigüedad media de las viviendas (años) |
| `total_rooms` | Total de habitaciones en el distrito |
| `total_bedrooms` | Total de dormitorios (**tiene valores faltantes**) |
| `population` | Población del distrito |
| `households` | Número de hogares |
| `median_income` | Ingreso medio de la zona (en decenas de miles de USD) |
| `median_house_value` | **Valor medio de las viviendas (USD)** ← nuestro objetivo |
| `ocean_proximity` | Cercanía al océano (variable **categórica**) |

### 6. Volvamos a nuestra pregunta

Recuerda las preguntas guía:

1. *¿El ingreso de una zona se relaciona con el valor de sus viviendas?*
2. *¿La cercanía al océano influye en el valor de las viviendas?*

Mirando las columnas, ¿cuáles crees que necesitaremos?

- Para la pregunta 1: `median_income` y `median_house_value`.
- Para la pregunta 2: `ocean_proximity` y `median_house_value`.

¡Ya tenemos identificadas las variables clave!
