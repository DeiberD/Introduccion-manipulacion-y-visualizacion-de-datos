## Seaborn: gráficas estadísticas fáciles y elegantes

**Seaborn** está construida sobre Matplotlib. Su ventaja es que produce **gráficas estadísticas
atractivas con muy poco código** y se integra perfectamente con los DataFrames de Pandas: le pasas
el DataFrame y le dices qué columnas usar.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")   # un estilo visual agradable por defecto
df = pd.read_csv("../data/housing.csv")
print("Dataset listo:", df.shape)
```

### 1. `histplot()` — histograma

El mismo histograma de antes, pero con Seaborn. Fíjate en lo poco que hay que escribir:

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="median_house_value", bins=50)
plt.title("Distribución del valor de las viviendas (Seaborn)")
plt.xlabel("Valor medio de la vivienda (USD)")
plt.ylabel("Número de distritos")
plt.show()
```

### 2. `scatterplot()` — dispersión con color por grupo

Aquí Seaborn brilla: con el parámetro **`hue`** podemos colorear los puntos según una variable
categórica. Veamos ingreso vs. valor, coloreando por cercanía al océano:

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="median_income", y="median_house_value",
                hue="ocean_proximity", alpha=0.4, s=15)
plt.title("Ingreso vs. valor de vivienda, por cercanía al océano")
plt.xlabel("Ingreso medio (decenas de miles de USD)")
plt.ylabel("Valor medio de la vivienda (USD)")
plt.show()
```

**¿Qué observamos?** Se confirma la tendencia ascendente (pregunta 1) **y además** vemos que
los puntos `INLAND` (tierra adentro) tienden a estar más abajo. Dos respuestas en una sola gráfica.

### 3. `boxplot()` — comparar grupos

El **diagrama de caja** (boxplot) resume la distribución de una variable para cada grupo: la línea
central es la mediana, la caja abarca el 50% central de los datos y los "bigotes" el resto.
Es perfecto para responder de forma contundente la **pregunta 2**:

```python
orden = ["INLAND", "NEAR OCEAN", "<1H OCEAN", "NEAR BAY", "ISLAND"]

plt.figure(figsize=(9, 6))
sns.boxplot(data=df, x="ocean_proximity", y="median_house_value", order=orden)
plt.title("Valor de la vivienda según la cercanía al océano")
plt.xlabel("Cercanía al océano")
plt.ylabel("Valor medio de la vivienda (USD)")
plt.xticks(rotation=15)
plt.show()
```

**¿Qué observamos?** Las viviendas `INLAND` tienen medianas claramente más bajas, mientras que
las cercanas al mar (y sobre todo `ISLAND`) son más caras. La respuesta a la pregunta 2 es
visualmente evidente: **sí, la cercanía al océano se asocia con mayor valor.**

### 4. `heatmap()` — ¿qué variable pesa más?

La **correlación** mide qué tan fuerte es la relación (lineal) entre dos variables numéricas, en una
escala de **-1 a 1**:

- Cerca de **1** → cuando una sube, la otra también.
- Cerca de **-1** → cuando una sube, la otra baja.
- Cerca de **0** → no hay relación lineal.

Pandas la calcula con `corr()` y Seaborn la convierte en un **mapa de calor** con una sola línea:

```python
correlaciones = df.select_dtypes("number").corr()   # solo columnas numéricas

plt.figure(figsize=(9, 7))
sns.heatmap(correlaciones, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlación entre las variables numéricas")
plt.show()
```

**¿Qué observamos?** Mira la fila de `median_house_value`: su correlación más alta, por lejos, es
con `median_income` (**0,69**). La antigüedad (`housing_median_age`, 0,11) y el número de
habitaciones (`total_rooms`, 0,13) apenas se relacionan con el valor.

**¿Cómo te fue con tu apuesta del inicio?** Entre las variables numéricas, el ingreso gana con
claridad. La cercanía al océano no aparece en este mapa porque es texto, pero su efecto ya lo vimos
en el boxplot.

### 5. Matplotlib vs. Seaborn: ¿cuándo usar cada uno?

| | Matplotlib | Seaborn |
|---|-----------|---------|
| **Código** | Más verboso, control total | Conciso, pensado para estadística |
| **Estética** | Básica por defecto | Bonita por defecto |
| **Integración con DataFrames** | Manual | Directa (`data=df`, `x=...`, `hue=...`) |
| **Ideal para** | Gráficas muy personalizadas | Exploración estadística rápida |

No son rivales: Seaborn **usa** Matplotlib por debajo, así que a menudo se combinan (de hecho,
seguimos usando `plt.title`, `plt.xlabel`, etc. sobre gráficas de Seaborn).
