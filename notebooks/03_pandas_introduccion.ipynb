## ¿Por qué Pandas?

NumPy es genial para números, pero los datos del mundo real suelen venir en **tablas**:
filas, columnas y con nombres. **Pandas** es la librería para trabajar con esas tablas
(cargarlas, explorarlas, filtrarlas, agruparlas...).

Pandas tiene dos estructuras principales:

- **`Series`** → una columna: datos de **una dimensión** con una etiqueta cada uno.
- **`DataFrame`** → una tabla completa: datos de **dos dimensiones** (filas × columnas).

### 1️⃣ Importar Pandas

Por convención se importa con el alias `pd`.

```python
import pandas as pd

print("Pandas versión:", pd.__version__)
```

### 2️⃣ La Series (una dimensión)

Una `Series` es como una columna con un índice. La creamos a partir de una lista:

```python
ingresos = pd.Series([8.32, 8.30, 7.25, 5.64, 3.85])
print(ingresos)
```

Fíjate que a la izquierda aparece el **índice** (0, 1, 2, ...) y a la derecha los **valores**.
No haremos mucho énfasis en las Series por ahora; lo importante es saber que **cada columna
de una tabla es una Series**.

### 3️⃣ El DataFrame (dos dimensiones)

Un `DataFrame` es una **tabla**. La forma más sencilla de crearlo es a partir de un
**diccionario**, donde cada clave es el nombre de una columna:

```python
datos = {
    "distrito":       ["A", "B", "C", "D"],
    "ingreso_medio":  [8.32, 5.64, 3.85, 2.10],
    "valor_vivienda": [452600, 341300, 189600, 91300],
    "cerca_oceano":   ["NEAR BAY", "NEAR BAY", "INLAND", "INLAND"],
}

df_ejemplo = pd.DataFrame(datos)
df_ejemplo
```

### 4️⃣ Anatomía de un DataFrame

Observa la tabla anterior e identifica sus partes:

- **Filas** → cada fila es un registro (aquí, un distrito). Se numeran con el **índice** (0, 1, 2, 3).
- **Columnas** → cada columna es una **variable** (`distrito`, `ingreso_medio`, ...).
- **Celdas** → la intersección de una fila y una columna: un valor concreto.

```python
print("Forma (filas, columnas):", df_ejemplo.shape)
print("Nombres de columnas    :", list(df_ejemplo.columns))
```

### 5️⃣ Seleccionar una columna = obtener una Series

```python
# Seleccionamos la columna 'valor_vivienda'
columna = df_ejemplo["valor_vivienda"]
print(columna)
print("\n¿Qué tipo es?", type(columna))   # ¡Una Series!
```
