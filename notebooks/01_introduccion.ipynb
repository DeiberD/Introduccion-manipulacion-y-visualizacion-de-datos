## ¿Qué vamos a hacer hoy?

En este taller vamos a recorrer el **flujo de trabajo básico del análisis de datos** con Python.
La idea es que, partiendo de un archivo de datos "crudo", lleguemos a **gráficas y conclusiones**
que nos ayuden a responder una pregunta concreta.

No necesitas experiencia previa con estas librerías. Vamos paso a paso.

## Primero: ¿qué es "analizar datos"?

Antes de escribir código, piensa un momento:

> **¿Qué entiendes tú por "analizar datos"?**

No hay una única respuesta correcta. En general, analizar datos es el proceso de **tomar información**,
**organizarla**, **hacerle preguntas** y **comunicar** lo que encontramos. Una definición práctica del flujo:

```
Cargar  →  Explorar  →  Manipular  →  Visualizar  →  Interpretar
```

Cada librería que veremos hoy cubre una parte de ese camino.

## Las cuatro librerías del taller

| Librería | ¿Para qué sirve? | ¿En qué parte del flujo aparece? |
|----------|------------------|----------------------------------|
| **NumPy** | Trabajar con **números y arreglos** de forma rápida y eficiente. Es la base numérica de casi todo el ecosistema. | Cálculo y operaciones numéricas |

| **Seaborn** | Crear **gráficas estadísticas** bonitas con muy poco código. Está construida sobre Matplotlib. | Visualizar |

No te preocupes por memorizar esto. Lo iremos viendo con ejemplos.
