Cargamos el dataset y las librerías de graficación:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../../data/housing.csv')
print('Dataset cargado:', df.shape)
```

### Ejercicio 1 - Histograma con Matplotlib

Crea un **histograma** de la variable `median_income` usando Matplotlib. Ponle título y nombres a los ejes.

```python
# Ejercicio 1
plt.figure(figsize=(8, 5))
plt.hist(df['median_income'], bins=40, color='teal', edgecolor='white')
plt.title('Distribución del ingreso medio')
plt.xlabel('Ingreso medio (decenas de miles de USD)')
plt.ylabel('Número de distritos')
plt.show()
```

### Ejercicio 2 - Dispersión con Matplotlib

Crea un **gráfico de dispersión** de `total_rooms` (eje X) contra `population` (eje Y). Usa transparencia (`alpha`) para manejar la gran cantidad de puntos. ¿Notas alguna relación?

```python
# Ejercicio 2
plt.figure(figsize=(8, 5))
plt.scatter(df['total_rooms'], df['population'], alpha=0.2, color='purple')
plt.title('Total de habitaciones vs. población del distrito')
plt.xlabel('Total de habitaciones')
plt.ylabel('Población')
plt.show()
```

### Ejercicio 3 - Histograma con Seaborn

Reproduce un histograma de `median_house_value` pero esta vez con **Seaborn** (`sns.histplot`). Usa el parámetro `bins=50`.

```python
# Ejercicio 3
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='median_house_value', bins=50)
plt.title('Distribución del valor de la vivienda (Seaborn)')
plt.xlabel('Valor medio de la vivienda (USD)')
plt.show()
```

### Ejercicio 4 - Boxplot por grupo

Crea un **boxplot** (`sns.boxplot`) que compare la distribución de `median_income` entre las distintas categorías de `ocean_proximity`. ¿En qué zonas la gente tiene más ingresos?

```python
# Ejercicio 4
plt.figure(figsize=(9, 6))
sns.boxplot(data=df, x='ocean_proximity', y='median_income')
plt.title('Ingreso medio según la cercanía al océano')
plt.xlabel('Cercanía al océano')
plt.ylabel('Ingreso medio')
plt.xticks(rotation=15)
plt.show()
```

### Ejercicio 5 - Conteo de categorías

Este gráfico no se vio en el taller. Usa `sns.countplot` para mostrar **cuántos distritos** hay en cada categoría de `ocean_proximity`.

```python
# Ejercicio 5
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='ocean_proximity')
plt.title('Número de distritos por cercanía al océano')
plt.xlabel('Cercanía al océano')
plt.ylabel('Cantidad de distritos')
plt.xticks(rotation=15)
plt.show()
```

### Ejercicio 6 - Mapa de calor de correlaciones

Un mapa de calor (heatmap) permite ver de un vistazo qué variables numéricas están relacionadas.

1. Calcula la matriz de correlación con `df.select_dtypes('number').corr()`.
2. Dibújala con `sns.heatmap(..., annot=True, cmap='coolwarm')`.

¿Qué variable tiene la correlación más alta con `median_house_value`?

```python
# Ejercicio 6
corr = df.select_dtypes('number').corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlación entre variables numéricas')
plt.show()

# La correlación más alta con median_house_value es median_income (~0.69).
```
