### Ejercicio 1 - Crear un arreglo y describirlo

Crea un arreglo de NumPy llamado `edades` con estos valores: 41, 21, 52, 52, 52, 42.
Luego imprime su `shape`, su `size` y su `ndim`.

````python
# Ejercicio 1
edades = np.array([41, 21, 52, 52, 52, 42])
print('shape:', edades.shape)
print('size :', edades.size)
print('ndim :', edades.ndim)
````

### Ejercicio 2 - Estadísticas básicas

Con el arreglo `edades` del ejercicio anterior, calcula e imprime: la **suma**, el **promedio**, el valor **máximo** y el **mínimo**.

````python
# Ejercicio 2
print('Suma    :', edades.sum())
print('Promedio:', edades.mean())
print('Máximo  :', edades.max())
print('Mínimo  :', edades.min())
````

### Ejercicio 4 - Operaciones entre dos arreglos

Tienes el total de habitaciones y el número de hogares de 4 distritos:

```python
habitaciones = np.array([880, 7099, 1467, 1274])
hogares      = np.array([126, 1138,  177,  219])
```

Calcula el **número de habitaciones por hogar** de cada distrito (divide un arreglo por el otro).

````python
# Ejercicio 4
habitaciones = np.array([880, 7099, 1467, 1274])
hogares      = np.array([126, 1138,  177,  219])
hab_por_hogar = habitaciones / hogares
print(hab_por_hogar.round(2))
````

### Ejercicio 5 - Máscaras booleanas

Este tema va un poco más allá del taller. Con el arreglo `valores = np.array([452600, 358500, 189600, 91300, 500001, 145000])`:

1. Calcula el promedio.
2. Usa una **condición** (`valores > promedio`) para quedarte solo con los valores por encima del promedio.

*Pista: en NumPy puedes hacer `valores[valores > umbral]`.*

````python
# Ejercicio 5
valores = np.array([452600, 358500, 189600, 91300, 500001, 145000])
promedio = valores.mean()
print('Promedio:', promedio)
print('Por encima del promedio:', valores[valores > promedio])
````

### Ejercicio 6 - Arreglos de dos dimensiones

Crea la siguiente matriz 2D (2 filas × 3 columnas):

```python
matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])
```

1. Imprime su `shape` y su `ndim`.
2. Calcula el promedio de **cada columna** usando `matriz.mean(axis=0)`.
3. Calcula el promedio de **cada fila** usando `matriz.mean(axis=1)`.

````python
# Ejercicio 6
matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])
print('shape:', matriz.shape, '| ndim:', matriz.ndim)
print('Promedio por columna (axis=0):', matriz.mean(axis=0))
print('Promedio por fila    (axis=1):', matriz.mean(axis=1))
````
