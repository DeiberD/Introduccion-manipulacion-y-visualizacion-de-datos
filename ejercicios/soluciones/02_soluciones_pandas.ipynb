### Ejercicio 4 - Filtrar con dos condiciones

Encuentra los distritos que están **tierra adentro** (`ocean_proximity == 'INLAND'`) **y** cuyo valor de vivienda supera los 200.000 USD. ¿Cuántos hay?

*Pista: combina condiciones con `&` y encierra cada una entre paréntesis.*

### Ejercicio 8 - Rellenar nulos

Este tema va más allá del taller. La columna `total_bedrooms` tiene valores faltantes.

1. Calcula la **mediana** de `total_bedrooms` (usa `.median()`).
2. Rellena los nulos con esa mediana usando `fillna(...)`.
3. Verifica que ya no queden nulos en esa columna.

*Pista: `df['total_bedrooms'] = df['total_bedrooms'].fillna(valor)`.*

```python
# Ejercicio 8
mediana = df['total_bedrooms'].median()
print('Mediana de total_bedrooms:', mediana)
df['total_bedrooms'] = df['total_bedrooms'].fillna(mediana)
print('Nulos restantes en total_bedrooms:', df['total_bedrooms'].isnull().sum())
```
